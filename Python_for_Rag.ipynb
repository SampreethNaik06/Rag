List Compprehension

```python
squares = []
for x in range(10):
  squares.append(x**2)
print(f"squares are : {squares}")
```

```python
# even number squares
squares = []
for x in range(10):
  if x % 2 ==0 :
    squares.append(x**2)
print(f"squares are : {squares}")
```

```python
# the upper example is without list comprehension  and the next one is with list conprehension

squares = [(x**2)for x in range(10)]
print(f"squares are : {squares}")

squares = [(x**2)for x in range(10) if x % 2 ==0 ]
print(f"squares are : {squares}")
```
